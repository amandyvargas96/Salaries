```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("http://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/Salaries.csv")
```

```python
df.head()
```

```python
# Calculate the average salary
df['salary'] .mean()
```

```python
# Calculate the median salary
df['salary'].median()
```

```python
# calculate the highest salary
df['salary'].max()

# Calculate the lowest salary
df['salary'].min()
```

```python
# calculate the highest salary
print("Higest salary:", df['salary'].max())

# Calculate the lowest salary
print("Lowest salary:", df['salary'].min())

```

```python
# Calculate average salary by rank
df.groupby('rank')['salary'].mean()
```

```python
# Calculate average salary by gender
df.groupby('sex')['salary'].mean()
```

```python
# Calculate correlation between years of service

df['yrs.service'].corr(df['salary'])
```

```python
 import matplotlib.pyplot as plt
 plt.scatter(df['yrs.service'], df['salary'])
 plt.xlabel('Years of Service')
 plt.ylabel('Salary')
 plt.title('Years of Service vs Salary')

 plt.show()
```
